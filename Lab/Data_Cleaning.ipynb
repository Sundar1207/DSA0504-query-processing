```python
# modules we'll use
import pandas as pd
import numpy as np

# read in all our data
df = pd.read_csv("data_cleaning.csv")

print(df.head(10))
# set seed for reproducibility
np.random.seed(0) 
```

Cleaning Techniques
1. Missing Values

```python
# Fill numeric columns
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# Fill categorical columns
df["Department"] = df["Department"].fillna(df["Department"].mode()[0])
df["Name"] = df["Name"].fillna("Unknown")

# Fill dates
df["Joining_Date"] = df["Joining_Date"].fillna(method="ffill")
df.head(10)
```

2. Scaling & Normalization

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df["Salary"] = scaler.fit_transform(df[["Salary"]])
df.head(10)
```

3. Date Cleaning

```python

df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    format="mixed",
    errors="coerce"
)

#Replace invalid dates

df["Joining_Date"] = df["Joining_Date"].fillna(df["Joining_Date"].mode()[0])
df.head(10)
```

4. Character Encoding

```python
df = pd.read_csv("C:\Documents\Query_processing\data_cleaning.csv", encoding="utf-8")

df = pd.read_csv("C:\Documents\Query_processing\data_cleaning.csv", encoding="latin1")

#Remove special characters

df["Name"] = (
    df["Name"]
      .str.encode("ascii", errors="ignore")
      .str.decode("ascii")
)
df.head(10)
```

5. Inconsistent Data Entry

```python
df["Department"] = df["Department"].str.strip()

#Convert to proper case

df["Department"] = df["Department"].str.title()

#Replace abbreviations

df["Department"] = df["Department"].replace({
    "Fin.":"Finance",
    "Hr":"HR",
    "Hr":"HR",
    "hr":"HR",
    "hR":"HR",
    "It":"IT",
    "it":"IT"
})

#Remove duplicate rows

df = df.drop_duplicates()

#Remove duplicate employee IDs

df = df.drop_duplicates(subset="Employee_ID")

df
```
