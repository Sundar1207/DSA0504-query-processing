```python
import pandas as pd

# Monthly Sales Data
data = {
    "Month": ["January", "February", "March", "April", "May", "June"],
    "Sales": [45000, 52000, 48000, 61000, 55000, 67000]
}

# Create DataFrame
df = pd.DataFrame(data)

print("Monthly Sales Data")
print(df)

print("\nData Aggregation")

# Total Sales
print("Total Sales:", df["Sales"].sum())

# Average Sales
print("Average Sales:", df["Sales"].mean())

# Maximum Sales
print("Maximum Sales:", df["Sales"].max())

# Minimum Sales
print("Minimum Sales:", df["Sales"].min())
```

```python
import pandas as pd

# Employee Data
data = {
    "Employee_ID": [101,102,103,104,105,106,107,108],
    "Employee_Name": [
        "John","Alice","David","Mary",
        "James","Robert","Linda","Sophia"
    ],
    "Department": [
        "HR","IT","IT","Finance",
        "HR","Finance","IT","HR"
    ],
    "Salary": [
        35000,50000,55000,60000,
        40000,65000,52000,45000
    ]
}

df = pd.DataFrame(data)

print("Employee Details")
print(df)

# Group by Department
result = df.groupby("Department").agg(
    Total_Salary=("Salary","sum"),
    Average_Salary=("Salary","mean"),
    Employee_Count=("Salary","count")
)

print("\nDepartment-wise Salary Details")
print(result)
```

```python
import pandas as pd

# Employees Data
employees = {
    "Employee_ID": [101,102,103,104,105],
    "Employee_Name": [
        "John","Alice","David","Mary","James"
    ],
    "Department_ID": [10,20,30,20,10],
    "Salary": [35000,50000,45000,60000,40000]
}

# Departments Data
departments = {
    "Department_ID": [10,20,30],
    "Department_Name": [
        "HR",
        "IT",
        "Finance"
    ]
}

# Create DataFrames
emp_df = pd.DataFrame(employees)
dept_df = pd.DataFrame(departments)

print("Employees Data")
print(emp_df)

print("\nDepartments Data")
print(dept_df)

# Merge DataFrames
merged_df = pd.merge(emp_df, dept_df, on="Department_ID")

print("\nMerged Data")
print(merged_df)
```
